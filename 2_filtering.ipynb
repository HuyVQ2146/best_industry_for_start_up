### Import

```python
import kagglehub
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Load data

```python
path = kagglehub.dataset_download("pranshuljoshi/global-startup-funding-and-valuation-cleaned-dataset")
file_path = path + "\\startup_data.csv"
df = pd.read_csv(file_path)
df
```

### Comparision

```python
df.groupby("Industry")[[
    "Employees", "Funding Amount (M USD)", "Valuation (M USD)", "Revenue (M USD)", "Market Share (%)", "Year Founded"
]].mean()
```

```python
df.groupby("Industry")[[
    "Employees", "Funding Amount (M USD)", "Valuation (M USD)", "Revenue (M USD)", "Market Share (%)", "Year Founded"
]].agg(["std", "min", "max"])
```

```python
df[((df["Industry"] == "E-Commerce") | (df["Industry"] == "IoT"))].groupby("Industry")[[
    "Employees", "Funding Amount (M USD)", "Valuation (M USD)", "Revenue (M USD)", "Market Share (%)", "Year Founded"
]].mean()
```

```python
employee = df.groupby("Industry")["Employees"].mean().sort_values(ascending=False)
colors = ['#1f77b4' if (industry == "IoT" or industry == "E-Commerce") else '#aed6f1' for industry in employee.index]

ax1 = employee.plot(kind="bar", figsize=(10, 6), title="Average Number of Employees by Industry", color=colors)
ax1.set_xlabel("Industry")
ax1.set_ylabel("Average Number of Employees")
plt.xticks(rotation=0)
plt.show()
```

```python
funding_amount = df.groupby("Industry")["Funding Amount (M USD)"].mean().sort_values(ascending=False)
colors = ['#1f77b4' if (industry == "IoT" or industry == "E-Commerce") else '#aed6f1' for industry in funding_amount.index]

ax1 = funding_amount.plot(kind="bar", figsize=(10, 6), title="Average Funding Amount by Industry", color=colors)
ax1.set_xlabel("Industry")
ax1.set_ylabel("Average Funding Amount (M USD)")
plt.xticks(rotation=0)
plt.show()
```

```python
valuation = df.groupby("Industry")["Valuation (M USD)"].mean().sort_values(ascending=False)
colors = ['#1f77b4' if (industry == "IoT" or industry == "E-Commerce") else '#aed6f1' for industry in valuation.index]

ax1 = valuation.plot(kind="bar", figsize=(10, 6), title="Average Valuation by Industry", color=colors)
ax1.set_xlabel("Industry")
ax1.set_ylabel("Average Valuation (M USD)")
plt.xticks(rotation=0)
plt.show()
```

```python
revenue = df.groupby("Industry")["Revenue (M USD)"].mean().sort_values(ascending=False)
colors = ['#1f77b4' if (industry == "IoT" or industry == "E-Commerce") else '#aed6f1' for industry in revenue.index]

ax1 = revenue.plot(kind="bar", figsize=(10, 6), title="Average Revenue by Industry", color=colors)
ax1.set_xlabel("Industry")
ax1.set_ylabel("Average Revenue (M USD)")
plt.xticks(rotation=0)
plt.show()
```

```python
market_share = df.groupby("Industry")["Market Share (%)"].mean().sort_values(ascending=False)
colors = ['#1f77b4' if (industry == "IoT" or industry == "E-Commerce") else '#aed6f1' for industry in market_share.index]

ax1 = market_share.plot(kind="bar", figsize=(10, 6), title="Average Market Share by Industry", color=colors)
ax1.set_xlabel("Industry")
ax1.set_ylabel("Average Market Share (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
year = df.groupby("Industry")["Year Founded"].mean().sort_values(ascending=False)
colors = ['#1f77b4' if (industry == "IoT" or industry == "E-Commerce") else '#aed6f1' for industry in year.index]

plt.ylim(2000, 2025)

ax1 = year.plot(kind="bar", figsize=(10, 6), title="Average Year Founded by Industry", color=colors)
ax1.set_xlabel("Industry")
ax1.set_ylabel("Average Year Founded")
plt.xticks(rotation=0)
plt.show()
```
